```python
# 딕셔너리는 어떻게 만들까?

# {key1: value1, key2: value2, ...}
dic = {'name' : 'sj',
	   'phone' : '01038031620',
	   'birth' : '1005'}
print(dic)
print(type(dic))
```

```python
a= 1
a= 2
print(a)
```

```python
# 딕셔너리의 키값으로 올 수 없는 데이터 타입은 리스트 -> mutable

a= [1,2,3]
a[0] = 3
print (a)
```

```python
# 딕셔너리 쌍 추가하기

a= {1:"a"}
a[2] = 'b'
print(a)
```

```python
a[3] = 'c'
print(a)
```

```python
a[4] = [1,2,3]
print(a)
```

```python
# 딕셔너리 요소 삭제하기

print(a)
del a[1]
print(a)
```

```python
# 딕셔너리에 key를 사용해서 value 얻기

grade = {
	'pey' : 10,
	'juli' : 99
}
print(grade['pey'])
print(grade['juli'])
```

```python
# 딕셔너리 관련 함수
## 1. key 리스트 만들기 - keys

a= {'name' : 'sj',
	   'phone' : '01038031620',
	   'birth' : '1005'
	   }
list(a.keys())
```

```python
## 2. value 리스트 만들기 - values

print(a.values())
list(a.values())
```

```python
## 3. key, value 쌍 얻기 - items

print(a)
```

```python
a.items()
```

```python
list(a.items())
```

```python
# key, value 쌍 모두 지우기 - clear

a.clear()
print(a)
```

```python
# key로 value 얻기 - get

a= {'name' : 'sj',
	   'phone' : '01038031620',
	   'birth' : '1005'
	   }
a['name']
```

```python
a.get('name')
```

```python
#a['hobby']
a.get('hobby',0)
```

```python
# 해당 key가 딕셔너리 안에 있는지 조사하기 - in

a= {'name' : 'sj',
	   'phone' : '01038031620',
	   'birth' : '1005'
	   }
'name' in a
```

```python
'email' in a
```

```python
# key로 value 꺼내기 - pop


a= {'name' : 'sj',
	   'phone' : '01038031620',
	   'birth' : '1005'
	   }

result = a.pop('name')
print(result)
print(a)
```
