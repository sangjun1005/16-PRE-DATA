```python
# 문자열은 어떻게 만들고 사용할까?

## 1. 큰 따옴표로 양쪽 둘러싸기

a = "hello world"

## 2. 작은 따옴표로 둘러싸기
b = 'python is fun'

## 3. 큰 따옴표 3개로 둘러싸기
c = """Life is too short, You need python"""

## 4. 작은 따옴표 3개로 둘러싸기
d = '''Life is too short, You need python'''

print(a)
print(b)
print(c)
print(d)
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```

```python
# 문자열 안에 작은 따옴표 포함하기

a = "python's favorite food is perl"
b = """python's favorite food is perl"""
c = 'python\'s favorite food is perl'
print(c)
```

```python
# 문자열 안에 큰 따온표 포함하기

b = "\"python is very easy\" he says."
a = 'python is very easy" he says.'

print(b)
```

```python
multiline = """Life is too short, 
You need python"""
multiline2 = "Life is too short, \nYou need python"

print(multiline2)
print(multiline)
```

```python
# 이스케이프 코드
#\n -> 문자열 안에서 줄을 바꿀 떄
#\t -> 문자열 안에서 간격을 둘 때
#\\ -> 문자열 안에서 열슬래시를 표현할 떄
#\' -> 문자열 안에서 작은따옴표 표현할 때
#\" -> 문자열 안에서 큰따옴표 표현할 때
```
